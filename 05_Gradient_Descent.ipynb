## Gradient Descent for Quadratic Function

**Aim:** Minimise f(x) = x² with gradient descent and visualise convergence.

**Instructions:** Run the cells in sequence. Record the output/plots, observations, result, and answers to the viva questions in your practical journal.

## Program

```python
import numpy as np
import matplotlib.pyplot as plt

def gd(start, lr, n=30):
    x=start; hist=[x]
    for _ in range(n):
        x=x-lr*(2*x); hist.append(x)
    return np.array(hist)

z=np.linspace(-10,10,400)
for lr in [.01,.1,.5,1.1]:
    h=gd(8,lr); plt.plot(h,h**2,'o-',label=f'lr={lr}')
plt.plot(z,z**2,'k--',label='f(x)=x²'); plt.ylim(-1,70)
plt.grid(); plt.legend(); plt.xlabel('x'); plt.ylabel('f(x)'); plt.show()
```
