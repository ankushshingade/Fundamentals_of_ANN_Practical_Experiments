**Level:** Beginner ANN mathematics and beginner Python

## Aim
To calculate the ReLU activation function for negative and positive values.

## Why this experiment is needed
ReLU removes negative values by replacing them with zero, but keeps positive values unchanged. It is widely used in hidden layers because of this simple behaviour.

## Mathematical idea

\[ReLU(z) = max(0, z)\]

## Python program
Every executable line or logical block has a comment explaining its purpose.

```python
# Define the ReLU activation function.
def relu(z):
    # max returns the larger of zero and z.
    return max(0.0, z)
# Store input values with negative, zero, and positive cases.
raw_outputs = [-5.2, -1.0, 0.0, 3.4, 8.1]
# Print headings for the output table.
print("Input z | ReLU output")
print("---------------------")
# Apply ReLU to every value.
for z in raw_outputs:
    # Print the original input and transformed result.
    print(f"{z:7.1f} | {relu(z):11.1f}")
```
