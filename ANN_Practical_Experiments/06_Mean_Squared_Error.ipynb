# Experiment 6: Mean Squared Error

**Level:** Beginner ANN mathematics and beginner Python

## Aim
To measure prediction error using Mean Squared Error.

## Why this experiment is needed
A network needs a numerical measure of how wrong its predictions are. MSE squares each error so negative and positive errors do not cancel, then averages them.

## Mathematical idea

\[MSE = sum((actual - prediction)^2) / n\]

## Python program

```python
# Store correct target values.
actual_targets = [1.0, 0.0, 1.0, 1.0]
# Store corresponding model predictions.
predictions = [0.8, 0.4, 0.9, 0.1]
# Start total squared error at zero.
total_squared_error = 0.0
# Find number of target-prediction pairs.
n = len(actual_targets)
# Visit every pair using its index.
for i in range(n):
    # Find signed error for one prediction.
    error = actual_targets[i] - predictions[i]
    # Square error so it is non-negative.
    squared_error = error ** 2
    # Add to the total.
    total_squared_error += squared_error
    # Show individual calculation.
    print(f"Item {i+1}: error={error:.2f}, squared error={squared_error:.2f}")
# Divide total by number of items to calculate MSE.
mse = total_squared_error / n
# Print final loss value.
print("MSE =", round(mse, 4))
```
