# Experiment 7: Iterative Weight Adjustment

**Level:** Beginner ANN mathematics and beginner Python

## Aim
To observe weight updates reducing prediction error.

## Why this experiment is needed
Learning means changing a weight after observing error. This simplified example has one input and one weight, allowing students to see every update without calculus.

## Mathematical idea

\[prediction = x*w; error = target - prediction; weight = weight + error*learning_rate*x\]

## Python program
Every executable line or logical block has a comment explaining its purpose.

```python
# Set one fixed input value.
x = 3.0
# Set the correct output that the model should learn.
target = 12.0
# Begin with an intentionally poor weight.
weight = 0.5
# Choose a small step size for each update.
learning_rate = 0.05
# Print column headings.
print("Epoch | Weight | Prediction | Error")
print("-----------------------------------")
# Repeat update process eight times.
for epoch in range(1, 9):
    # Calculate current model prediction.
    prediction = x * weight
    # Find difference between target and prediction.
    error = target - prediction
    # Update weight in a direction that reduces error.
    weight = weight + error * learning_rate * x
    # Display learning progress.
    print(f"{epoch:5d} | {weight:6.3f} | {prediction:10.3f} | {error:5.2f}")
```
