## Program

```python
import torch, torch.nn as nn, matplotlib.pyplot as plt, seaborn as sns
from sklearn.datasets import make_classification
from sklearn.model_selection import train_test_split
from sklearn.metrics import confusion_matrix
X,y=make_classification(n_samples=1000,n_features=10,n_informative=8,n_classes=4,random_state=42)
Xt,Xv,yt,yv=train_test_split(X,y,test_size=.2,random_state=42)
Xt=torch.tensor(Xt,dtype=torch.float32); Xv=torch.tensor(Xv,dtype=torch.float32)
yt=torch.tensor(yt); yv=torch.tensor(yv)
model=nn.Sequential(nn.Linear(10,32),nn.ReLU(),nn.Linear(32,4))
opt=torch.optim.Adam(model.parameters(),lr=.001); lossfn=nn.CrossEntropyLoss()
for _ in range(300):
    loss=lossfn(model(Xt),yt); opt.zero_grad(); loss.backward(); opt.step()
pred=model(Xv).argmax(1); cm=confusion_matrix(yv,pred)
sns.heatmap(cm,annot=True,fmt='d',cmap='Blues'); plt.xlabel('Predicted'); plt.ylabel('True'); plt.show()
print('Test accuracy:',(pred==yv).float().mean().item())
```
