**Level:** Beginner ANN mathematics and beginner Python

## Aim
To convert a numerical neuron score into a binary decision.

## Why this experiment is needed
A weighted sum can be any number. A step activation makes a hard decision: 1 for a score at or above threshold, otherwise 0.

## Mathematical idea

\[y = 1 if z >= threshold; otherwise y = 0\]

## Python program

```python
# Define a reusable function named step_function.
def step_function(score, threshold=0.0):
    # Compare score with the selected threshold.
    if score >= threshold:
        # Return 1 for a positive decision.
        return 1
    else:
        # Return 0 for a negative decision.
        return 0
# Create several scores to test the function.
test_scores = [-3.5, -0.1, 0.0, 0.5, 4.2]
# Print headings for a readable table.
print("Score | Step output")
print("-------------------")
# Test every score in the list.
for score in test_scores:
    # Call the function and display its binary result.
    print(f"{score:5.1f} | {step_function(score)}")
```
