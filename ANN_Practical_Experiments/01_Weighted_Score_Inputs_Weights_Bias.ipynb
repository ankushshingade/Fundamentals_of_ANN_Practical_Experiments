# Experiment 1: Weighted Score Inputs Weights Bias

**Level:** Beginner ANN mathematics and beginner Python

## Aim
To calculate the weighted sum of inputs and bias for one artificial neuron.

## Why this experiment is needed
A neuron receives several inputs. Each input is multiplied by its weight, the products are added, and a bias is added. This experiment connects the formula with ordinary Python lists and a loop.

## Mathematical idea

\[z = sum(x_i*w_i) + b\]

## Python program
Every executable line or logical block has a comment explaining its purpose.

```python
# Store three input features in a Python list.
inputs = [4.0, 0.9, 7.0]
# Store one matching importance value (weight) for every input.
weights = [1.5, 2.0, 0.5]
# Bias is a constant added after all weighted products are summed.
bias = -5.0
# Start total at zero before adding each product.
weighted_sum = 0.0
# range(len(inputs)) gives positions 0, 1, and 2.
for i in range(len(inputs)):
    # Multiply one input by its matching weight.
    product = inputs[i] * weights[i]
    # Add this product to the running total.
    weighted_sum = weighted_sum + product
    # Print this one step so the calculation is visible.
    print(f"Input {i+1}: {inputs[i]} x {weights[i]} = {product:.3f}")
# Add bias to obtain the final linear neuron value z.
z = weighted_sum + bias
# Print the intermediate and final answers.
print("Weighted sum =", weighted_sum)
print("Final neuron value z =", z)
```
