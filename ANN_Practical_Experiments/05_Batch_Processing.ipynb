# Experiment 5: Batch Processing

**Level:** Beginner ANN mathematics and beginner Python

## Aim
To process several input samples using the same weights and bias.

## Why this experiment is needed
A trained neuron uses the same weights and bias for every sample. This experiment demonstrates a small batch with lists and loops before introducing NumPy matrices.

## Mathematical idea

\[For each sample: z = x1*w1 + x2*w2 + b\]

## Python program
Every executable line or logical block has a comment explaining its purpose.

```python
# Create a batch: each inner list is one sample with two features.
batch_X = [[1.0, 2.0], [3.0, 0.5], [-1.0, 4.0]]
# Use the same two weights for every sample.
weights = [0.4, -0.6]
# Use one common bias for every sample.
bias = 0.5
# enumerate gives both sample number and sample values.
for index, x in enumerate(batch_X):
    # Multiply first feature by first weight.
    first_part = x[0] * weights[0]
    # Multiply second feature by second weight.
    second_part = x[1] * weights[1]
    # Add both products and bias.
    z = first_part + second_part + bias
    # Display score for this sample.
    print(f"Sample {index+1} {x} -> z = {z:.2f}")
```
