# Experiment 10: Two Layer XOR Network

**Level:** Beginner ANN mathematics and beginner Python

## Aim
To trace forward propagation through a two-layer network for XOR.

## Why this experiment is needed
Hidden neurons create intermediate features. Their outputs are combined by an output neuron, allowing the network to solve a problem that one neuron cannot solve.

## Mathematical idea

\[Hidden layer -> output layer; each neuron uses weighted sum followed by activation.\]

## Python program

```python
# Import math for the exponential function.
import math
# Define Sigmoid for every neuron in this example.
def sigmoid(z):
    # Convert any score into a value between 0 and 1.
    return 1.0 / (1.0 + math.exp(-z))
# Define a function for one complete forward pass.
def forward_pass_network(x1, x2):
    # Calculate hidden neuron 1 score and activation.
    h1 = sigmoid(x1*1.0 + x2*1.0 - 0.5)
    # Calculate hidden neuron 2 score and activation.
    h2 = sigmoid(x1*1.0 + x2*1.0 - 1.5)
    # Combine hidden outputs in the final output neuron.
    output = sigmoid(h1*2.0 + h2*(-2.0) - 1.0)
    # Return hidden values and final probability.
    return h1, h2, output
# Test all four XOR input pairs.
for x1, x2 in [(0,0), (0,1), (1,0), (1,1)]:
    # Run one forward pass.
    h1, h2, probability = forward_pass_network(x1, x2)
    # Convert probability into final binary prediction.
    prediction = 1 if probability >= 0.5 else 0
    # Print all stages for inspection.
    print(f"({x1},{x2}) -> h1={h1:.3f}, h2={h2:.3f}, output={probability:.3f}, prediction={prediction}")
```
