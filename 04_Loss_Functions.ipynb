## Program

```python
import numpy as np

def mse(y, p): return np.mean((y-p)**2)
def bce(y, p):
    p=np.clip(p,1e-12,1-1e-12)
    return -np.mean(y*np.log(p)+(1-y)*np.log(1-p))

y_reg=np.array([1.,2.,3.,4.])
print('Good regression MSE:',mse(y_reg,np.array([1.1,1.9,3.2,3.8])))
y=np.array([1,0,1,1,0])
for name,pred in [('Confident correct',[.95,.05,.9,.85,.1]),('Uncertain',[.55,.45,.6,.58,.4]),('Wrong confident',[.1,.9,.15,.2,.85])]:
    print(name, 'BCE =', bce(y,np.array(pred)))
```
