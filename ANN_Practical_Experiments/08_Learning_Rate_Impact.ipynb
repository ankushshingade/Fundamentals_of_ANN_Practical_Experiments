# Experiment 8: Learning Rate Impact

**Level:** Beginner ANN mathematics and beginner Python

## Aim
To compare slow, suitable, and excessively large learning rates.

## Why this experiment is needed
Learning rate controls update size. Too small means slow learning; too large can overshoot a correct value. Students should observe this before using optimisers.

## Mathematical idea

\[weight = weight + error*learning_rate\]

## Python program

```python
# Create a function so the same experiment can run with different learning rates.
def train(lr):
    # Use fixed input and target for a fair comparison.
    x, target = 2.0, 10.0
    # Begin every run from the same starting weight.
    weight = 0.0
    # Print selected learning rate.
    print(f"\nLearning rate = {lr}")
    # Perform three learning updates.
    for epoch in range(1, 4):
        # Calculate current prediction.
        prediction = x * weight
        # Find current error.
        error = target - prediction
        # Update weight using selected rate.
        weight = weight + error * lr
        # Print state before next epoch.
        print(f"Epoch {epoch}: weight={weight:.2f}, prediction={prediction:.2f}")
# Run a slow-learning case.
train(0.01)
# Run a moderate case.
train(0.25)
# Run an overshooting case.
train(1.10)
```
