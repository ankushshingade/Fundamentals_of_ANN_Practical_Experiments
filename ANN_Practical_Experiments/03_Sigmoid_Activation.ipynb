# Experiment 3: Sigmoid Activation

**Level:** Beginner ANN mathematics and beginner Python

## Aim
To calculate Sigmoid outputs and understand smooth probability-like mapping.

## Why this experiment is needed
Unlike the step function, Sigmoid changes gradually. It maps every real number to a value between 0 and 1, so small changes in score create small changes in output.

## Mathematical idea

\[sigmoid(z) = 1 / (1 + e^(-z))\]

## Python program
Every executable line or logical block has a comment explaining its purpose.

```python
# Import math so that exp() can calculate e raised to a power.
import math
# Define the Sigmoid activation function.
def sigmoid(z):
    # Return a number strictly between 0 and 1.
    return 1.0 / (1.0 + math.exp(-z))
# Select scores below, at, and above zero.
scores = [-6.0, -2.0, 0.0, 2.0, 6.0]
# Print a table heading.
print("Score z | Sigmoid output")
print("------------------------")
# Calculate one Sigmoid output for each score.
for z in scores:
    # Call the function and store the result.
    probability = sigmoid(z)
    # Display four decimal places and a percentage.
    print(f"{z:7.1f} | {probability:.4f} ({probability*100:.1f}%)")
```
