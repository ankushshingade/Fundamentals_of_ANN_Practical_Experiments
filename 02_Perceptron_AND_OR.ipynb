## Perceptron for AND and OR Gates

**Aim:** Implement a single perceptron from scratch and train it for AND and OR logic gates.

## Program

```python
import numpy as np

class Perceptron:
    def __init__(self, lr=0.1, epochs=30): self.lr, self.epochs = lr, epochs
    def fit(self, X, y):
        self.w = np.zeros(X.shape[1]); self.b = 0.0
        for _ in range(self.epochs):
            for x, target in zip(X, y):
                pred = int(np.dot(x, self.w) + self.b >= 0)
                error = target - pred
                self.w += self.lr * error * x
                self.b += self.lr * error
    def predict(self, X): return (X @ self.w + self.b >= 0).astype(int)

X = np.array([[0,0],[0,1],[1,0],[1,1]])
for name, y in [('AND',np.array([0,0,0,1])), ('OR',np.array([0,1,1,1]))]:
    model = Perceptron(); model.fit(X, y)
    print(f'\n{name} gate')
    print('Weights:', model.w, ' Bias:', model.b)
    print('Predictions:', model.predict(X))
    print('Targets:    ', y)
```
