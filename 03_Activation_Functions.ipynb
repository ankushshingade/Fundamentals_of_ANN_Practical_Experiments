## Program

```python
import numpy as np
import matplotlib.pyplot as plt
x=np.linspace(-10,10,1000)
sig=lambda z:1/(1+np.exp(-z))
relu=lambda z:np.maximum(0,z)
lrelu=lambda z:np.where(z>0,z,0.01*z)
funcs=[('Sigmoid',sig(x),sig(x)*(1-sig(x))),('Tanh',np.tanh(x),1-np.tanh(x)**2),('ReLU',relu(x),(x>0).astype(float)),('Leaky ReLU',lrelu(x),np.where(x>0,1,.01))]
fig,ax=plt.subplots(4,2,figsize=(10,12))
for i,(name,y,dy) in enumerate(funcs):
    ax[i,0].plot(x,y); ax[i,0].set_title(name); ax[i,0].grid()
    ax[i,1].plot(x,dy); ax[i,1].set_title(name+' derivative'); ax[i,1].grid()
plt.tight_layout(); plt.show()
```
