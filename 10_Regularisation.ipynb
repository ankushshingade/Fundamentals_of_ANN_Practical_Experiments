## Dropout and L2 Regularisation

**Aim:** Apply dropout and L2 regularisation and compare train and validation loss.

## Program

```python
import torch, torch.nn as nn, matplotlib.pyplot as plt
torch.manual_seed(42)
X=torch.randn(300,10); y=(X[:,:1]*2+0.5*torch.randn(300,1))
Xt,Xv,yt,yv=X[:100],X[100:],y[:100],y[100:]
def train(drop=0.,wd=0.):
    model=nn.Sequential(nn.Linear(10,64),nn.ReLU(),nn.Dropout(drop),nn.Linear(64,64),nn.ReLU(),nn.Dropout(drop),nn.Linear(64,1))
    opt=torch.optim.Adam(model.parameters(),lr=.003,weight_decay=wd); tr=[]; va=[]
    for _ in range(250):
        model.train(); loss=nn.functional.mse_loss(model(Xt),yt); opt.zero_grad(); loss.backward(); opt.step()
        model.eval()
        with torch.no_grad():
            tr.append(nn.functional.mse_loss(model(Xt),yt).item())
            va.append(nn.functional.mse_loss(model(Xv),yv).item())
    return tr,va
for label,d,wd in [('Baseline',0,0),('Dropout',.4,0),('L2',0,.01)]:
    tr,va=train(d,wd); plt.plot(tr,label=label+' train'); plt.plot(va,'--',label=label+' validation')
plt.yscale('log'); plt.legend(ncol=2); plt.xlabel('Epoch'); plt.ylabel('MSE Loss'); plt.grid(); plt.show()
```
