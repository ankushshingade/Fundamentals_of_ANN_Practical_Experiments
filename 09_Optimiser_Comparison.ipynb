**Aim:** Compare SGD and Adam optimisers using selected learning rates.

## Program

```python
import torch, torch.nn as nn, matplotlib.pyplot as plt
torch.manual_seed(1); X=torch.randn(400,5); y=(X[:,0:1]*2-X[:,1:2]+.2*torch.randn(400,1))
def run(kind,lr):
    model=nn.Sequential(nn.Linear(5,16),nn.ReLU(),nn.Linear(16,1))
    opt=(torch.optim.SGD if kind=='SGD' else torch.optim.Adam)(model.parameters(),lr=lr)
    hist=[]
    for _ in range(150):
        loss=nn.functional.mse_loss(model(X),y); opt.zero_grad(); loss.backward(); opt.step(); hist.append(loss.item())
    return hist
for name,lr in [('SGD',.01),('SGD',.1),('Adam',.001),('Adam',.01)]:
    h=run(name,lr); plt.plot(h,label=f'{name}, lr={lr}'); print(name,lr,'final loss=',h[-1])
plt.yscale('log'); plt.legend(); plt.xlabel('Epoch'); plt.ylabel('MSE Loss'); plt.grid(); plt.show()
```
